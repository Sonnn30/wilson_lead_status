```python
df['Full Name'] = np.where(
    df['First Name'].isna() & df['Last Name'].isna(),  
    df['Full Name'],                                    
    df['First Name'].fillna('') + ' ' + df['Last Name'].fillna('')  
)
```

```python
df_titik = df[df['Full Name'].str.contains('.', regex=False, na=False)]
df_titik['Full Name'].unique()
```

```python
mask = df['Full Name'].str.contains('.', regex=False, na=False)
idx_list = df[mask].index.tolist()

for idx in idx_list:
    if idx == 0:
        print(f"Index {idx}: tidak punya baris sebelumnya, dilewati.")
        continue
    
    first_name_prev = None
    search_idx = idx - 1
    
    while search_idx >= 0:
        candidate = df.loc[search_idx, 'First Name']
        if pd.notna(candidate):
            first_name_prev = str(candidate)
            break
        search_idx -= 1 
    
    if first_name_prev is None:
        print(f"Index {idx}: tidak ada First Name valid di atasnya, dilewati.")
        continue
    
    full_name_current = df.loc[idx, 'Full Name']
    inisial = full_name_current.split('.')[0].strip() + '.'
    
    full_name_baru = full_name_current.replace(inisial, first_name_prev)
    df.loc[idx, 'Full Name'] = full_name_baru
    
    print(f"Index {idx}: '{full_name_current}' = '{full_name_baru}'")

print(df['Full Name'])
```

```python
import pandas as pd

df = pd.read_csv("data_clean.csv")

def standarisasi_tanggal(kolom):
    hasil = pd.Series([pd.NaT] * len(kolom), index=kolom.index)

    # Format 1: MM/DD/YYYY 
    fmt1 = pd.to_datetime(kolom, format='%m/%d/%Y', errors='coerce')
    hasil = hasil.where(fmt1.isna(), fmt1)

    # Format 2: YYYY-MM-DD 
    fmt2 = pd.to_datetime(kolom, format='%Y-%m-%d', errors='coerce')
    hasil = hasil.where(~hasil.isna() | fmt2.isna(), fmt2)

    # Format 3: YYYY-MM-DDTHH:MM:SSZ
    fmt3 = pd.to_datetime(kolom, format='%Y-%m-%dT%H:%M:%SZ', errors='coerce')
    hasil = hasil.where(~hasil.isna() | fmt3.isna(), fmt3)

    # Konversi ke format YYYY-MM-DD
    return hasil.dt.strftime('%Y-%m-%d')

df['Create Date']        = standarisasi_tanggal(df['Create Date'])
df['Last Modified Date'] = standarisasi_tanggal(df['Last Modified Date'])

print(df[['Create Date', 'Last Modified Date']].head(10))
print(f"\nCreate Date NaT    : {df['Create Date'].isna().sum()}")
print(f"Last Modified NaT  : {df['Last Modified Date'].isna().sum()}")
```
